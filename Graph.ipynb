```python
import networkx as nx
import matplotlib as plt
D=nx.Graph()
D.add_node('A')
D.add_node('B')
D.add_node('C')
D.add_node('D')
D.add_edge('A','B')
D.add_edge('A','C')
D.add_edge('B','C')
D.add_edge('C','D')
D.add_edge('D','A')
D.add_edge('D','B')

nx.draw(D,with_labels=True,node_size=1000,node_color)


import plotly.express as px
import plotly.io as pio
import pandas as pd
import matplotlib.pyplot as plt
pio.renderers.deafult='notebook'
```

```python
df=pd.read_csv('social_media_engagement.csv')
df
```

```python
fig=px.bar(df,x='post_type',y='likes',
           height=400,
           width=600,
           hover_data=['post_day','sentiment_score','post_time'],
           color='post_type',
           color_discrete_sequence=['red','blue','green'])
fig.show()
```
